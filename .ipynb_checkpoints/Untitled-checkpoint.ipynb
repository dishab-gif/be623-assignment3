```python
#store hbb DNA sequence
hbb = "ATGGTGCATCTGACTCCTGAGGAGAAGTCTGCCGTTACTGCCCTGTGGGGCAAGGTGAAC"

#Check data type
print(type(hbb))
print(type(len(hbb)))
print(type(56.67))
print(type(True))
```

```python
#print the first codon of sequence
print(hbb[0:3])

#print the second codon
print(hbb[3:6])

#print last 3 bases
print(hbb[-3:])

#calculate the complete codon
print(len(hbb)/3)
```

```python
#convert DNA sequence into RNA by replacing T with U
print(hbb.replace("T", "U"))

#convert T to U of last 12 bases
print(hbb.replace("T", "U")[-12:])

#convert T to U of last 12 bases (other way)
print(hbb[-12:].replace("T", "U"))
```

```python
#count number of G in hbb
print(hbb.count("G"))

#count number of GC patterns
print(hbb.count("GC"))

#calculate number of bases that are neither G or C
print(len(hbb) - hbb.count("G") - hbb.count("C"))
```

```python
#create list of accession numbers
accessions = ["NM_000518", "P68871", "1HHO"]

#calculate number of accession number in the list
print(len(accessions))

#count of the position from 1
count = 1

#print position and accession number
for acc in accessions:
    print(count,acc)
    count = count + 1
```

```python
accessions = ["NM_000518", "P68871", "1HHO"]
#add new accession number
accessions.append("ENSG00000244734")

#print length of the list
print(len(accessions))

#count of the position from 1
count = 1

#print position and accession number
for acc in accessions:
    print(count,acc)
    count = count + 1

#print last item of list
print(accessions[-1])
```

```python
#create dictionary mapping the accessions
databases = {
"NM_000518": "NCBI Nucleotide",
"P68871": "UniProt",
"1HHO": "PDB",
}
#print number of entries
print(len(databases))

#print value stored for it
print(databases["P68871"])
```

```python
#Print one line per entry, in the form accession -> database
for acc, db in databases.items():
    print(acc, "->", db)

#Ask for an accession that is not in the dictionary and make your code print not found
print(databases.get("XYZ", "Not found"))

#Test whether one of the database names is in the dictionary
print("PDB" in databases)
```

```python
#cresting an emplty list
headers = []

#read FASTA file line by line
for line in open("a3.fasta"):
    if line.startswith(">"):
        headers.append(line)

#print number of headers
print(len(headers))

#print each header
for header in headers:
    print(header)
    
```
